#### Newton method work to find value where the f(x) = 0  
#### as we need the minimum value of our function c(m,b) we require c'(m,b) = 0  because it tell the value where slope a=is zero that the point of minima or maxima 

```python
# REQUIREMENT

# define cost(m,b)
# start with some initial (m0,b0) 
# update them using  m1,b1 = (m0,b1) - grad(c(m,b)) / hessian(c(m,b))
# so , find gradient and hessian of c(m,b)
```

```python
from sympy import *
import numpy as np

m,b = symbols('m b')

tx = [2,4,6,8,9]
ty = [3,6,8,2,5]

cost = 0

for i in range(len(tx)):
    cost += (m*tx[i] + b - ty[i])**2

print(cost)
```

```python
# gradient calculation 

def gradient(cost):
    dm = diff(cost,m)
    db = diff(cost,b)
    return dm,db

```

```python
# hessian matrix

def hessian(dm,db):
    dmb = diff(dm,b)
    dmm = diff(dm,m)
    dbm = diff(db,m)
    dbb = diff(db,b)

    return np.array([[dmm,dmb],[dbm,dbb]],dtype=float)

```

```python
# now combined all to calculate the value
dm,db = gradient(cost)
hessian_f = hessian(dm,db)

def gradient_c(a_c):
    return np.array([dm.evalf(subs={m:a_c[0],b:a_c[1]}),db.evalf(subs={m:a_c[0],b:a_c[1]})])


def newton_optimization(a_c,grad,hess,iteration):
    for i in range(iteration):
        a_c = a_c - np.matmul(np.linalg.inv(hess),grad(a_c))
        print("m:",a_c[0],"b:",a_c[1])

    return a_c
```

```python
# lets check it 

a_c = np.array([2,3])

newton_optimization(a_c,gradient_c,hessian_f,10)
```
