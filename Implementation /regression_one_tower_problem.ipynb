```python
# cost function 

def cost(tx,ty,m,b):
    c = 0
    for i in range(0,np.size(tx)):
        c += (m*tx[i] + b - ty[i])**2

    return c




# print(np.array([expr.evalf(subs={m:2,b:3}) for expr in cost_exp]))
```

```python
# now calculate the gradient that use for gradient descent 

def gradient(m,b):
    dm = np.zeros(len(tx))
    db = np.zeros(len(tx))

    for i in range(len(tx)):
        error = m*tx[i] + b - ty[i]
        dm[i] = 2*error*tx[i]
        db[i] = 2*error

    return dm,db


dcdm , dcdb = gradient(2,3)
print(dcdm,dcdb)
```

```python
# now we have (tx,ty) coordinate for the towers and cost function  
# need to minimization of this cost function usign gradient descent method 

# we need to choose some learning rate = alpha , and number iteration we want and the gradient return a value which minimum/near to minimum 


def gradient_descent(dcdm,dcdb,m,b,alpha=0.2,iteration=100):

    for i in range(iteration):
        m , b = m + alpha*dcdm , b + alpha*dcdb
        print("m: "+ m , " b: " + b)

    return m,b
```
