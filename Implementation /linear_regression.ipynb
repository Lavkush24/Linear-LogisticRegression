### Lets Write general class Linear_regression for the finding minimum cost problem

```python
import numpy as np
from sympy import *
from sympy.utilities.lambdify import lambdify
```

```python
class LinearRegression:
    def __init__(self,x,y,data_point,m,b):
        self.x = x
        self.y = y
        self.m = m
        self.b = b
        self.data_point = data_point

    def cost(self):
        cost = 0
        for i in range(self.data_point):
            cost += (self.m*self.x[i] + self.b - self.y[i])**2

        return cost


    def gradient(self):
        cmb = self.cost()
        dm,db = diff(cmb,self.m), diff(cmb,self.b)
        return np.array([dm,db])


    def hessian(self):
        grad = self.gradient()
        dm = grad[0]
        db = grad[1]
        dmm = diff(dm,self.m)
        dmb = diff(dm,self.b)
        dbb = diff(db,self.b)
        dbm = diff(db,self.m)

        return np.array([[dmm,dmb],[dbm,dbb]],dtype=float)


    def gradient_descent(self,m_b,alpha,iteration):
        history = []
        grad = self.gradient()

        for i in range(iteration):
            dm = grad[0].evalf(subs={self.m:m_b[0],self.b:m_b[1]})
            db = grad[1].evalf(subs={self.m:m_b[0],self.b:m_b[1]})
            m_b[0],m_b[1] = m_b[0] - alpha*dm , m_b[1] - alpha*db
            history.append(m_b.copy())
            print(f"m: {m_b[0]} b:{m_b[1]}")

        return history

    def gradient_optimized(self,m_b,alpha,iteration,p):
            history = []
            grad = self.gradient()
            dm_f = lambdify((self.m,self.b),grad[0],'numpy')
            db_f = lambdify((self.m,self.b),grad[1],'numpy')
    
            for i in range(iteration):
                dm = dm_f(m_b[0],m_b[1])
                db = db_f(m_b[0],m_b[1])
                m_b[0],m_b[1] = m_b[0] - alpha*dm , m_b[1] - alpha*db
                history.append(m_b.copy())
                if i%p == 0:
                    print(f"m: {m_b[0]} b:{m_b[1]}")
            
            return history


    def newton_optimization(self,m_b,iteration):
        history = []
        hess = self.hessian()
        hess = np.linalg.inv(hess)
        grad_exp = self.gradient()

        for i in range(iteration):
            grad = [i.evalf(subs={self.m:m_b[0],self.b:m_b[1]}) for i in grad_exp]
            m_b = m_b - np.matmul(hess,grad)
            history.append(m_b.copy())
            print(f"m: {m_b[0]} b:{m_b[1]}")

        return history


        
```

```python
m,b = symbols('m b')
l1 = LinearRegression([2,4,6,8,9],[3,6,8,2,5],5,m,b)

l1.gradient_descent([2,3],0.5,10)
```
