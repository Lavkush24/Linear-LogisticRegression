```python
from sympy import *
import numpy as np

from sympy.utilities.lambdify import lambdify
```

```python
"""
some of common sympy function/methods 

sqrt()
N(num,decimalpalces)
symbols('x y ')
factor(expression)
expand(expression)
expression.evalf(subs{a:,b:}) to find value of function at some instace

"""


# x = sqrt(18)

# print(x)

# print(N(x,6))


x,y = symbols('x y')

expr = x**2 + x*y + y**3

expr -= (x**4+x*y+y**3)

factor(expr)
print(factor(expr))

print("f(3,4) = ", expr.evalf(subs={x:3,y:4}))


arry = np.array([1,2,3])
fn_sqr = x**x

fn_sqr_sym = lambdify(x, fn_sqr,'numpy')  # to make function numpy freindly
print(fn_sqr_sym(arry))


print(diff(x**3,x))
```

```python
a , b = symbols('a b',positive='Integer')
c = (x-a)**2  + (b-x)**2 

dfdx = diff(c,x)

print(dfdx)

f23 = dfdx.evalf(subs={a:2,b:3})

print(f23)
```
